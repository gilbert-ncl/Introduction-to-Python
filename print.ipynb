### 1. 基本輸出print()

`print()` 函數是 Python 中用於將內容輸出到控制台的常用函數。它可以輸出文字、變數的值，甚至多個項目。

**基本用法：**

您可以直接在 `print()` 函數的括號中放入您想要輸出的內容。

```python
name = "Alice"
print(name)
```

```python
age = 30
print("我的名字是", name, "，我今年", age, "歲。")
```

```python
city = "台北"
print(f"我住在{city}。")
```

```python
print("這行會換行")
print("這行不會換行", end="")
print("這行在同一行")
```

## 基本變數概念

變數是儲存值的容器。在Python中，您不需要宣告變數的類型，Python會根據您賦予的值自動判斷。

### 變數的賦值

使用等號 (`=`) 將值賦予變數。

```python
# 將整數 10 賦予變數 age
age = 10
# 將字串 "Alice" 賦予變數 name
name = "Alice"
# 將浮點數 3.14 賦予變數 pi
pi = 3.14
# 將布林值 True 賦予變數 is_student
is_student = True

print(f"age: {age}, type: {type(age)}")
print(f"name: {name}, type: {type(name)}")
print(f"pi: {pi}, type: {type(pi)}")
print(f"is_student: {is_student}, type: {type(is_student)}")
```

### 變數的重新賦值

您可以隨時改變變數的值。

```python
# 重新賦值
age = 11
name = "Bob"

print(f"新的 age: {age}")
print(f"新的 name: {name}")
```

### 變數命名規則

*   變數名稱可以包含字母、數字和底線 (`_`)。
*   變數名稱不能以數字開頭。
*   變數名稱不能使用Python的保留字（例如 `if`, `for`, `while` 等）。
*   變數名稱區分大小寫（`age` 和 `Age` 是不同的變數）。
